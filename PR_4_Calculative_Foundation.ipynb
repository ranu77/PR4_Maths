## Part A: Vector & Matrix Fundamentals

### 1. Student Score Vectors

```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
student_A = np.array([80, 75, 85])
student_B = np.array([65, 90, 78])

student_A, student_B
```

### 2. L1 and L2 Norms

```python
L1_A = np.linalg.norm(student_A, ord=1)
L2_A = np.linalg.norm(student_A, ord=2)

L1_B = np.linalg.norm(student_B, ord=1)
L2_B = np.linalg.norm(student_B, ord=2)

print("Student A L1 Norm:", L1_A)
print("Student A L2 Norm:", L2_A)
print("Student B L1 Norm:", L1_B)
print("Student B L2 Norm:", L2_B)
```

### 3. Dot Product

```python
dot_product = np.dot(student_A, student_B)

print("Dot Product:", dot_product)
```

### 4. Angle Between Two Students

```python
magnitude_A = np.linalg.norm(student_A)
magnitude_B = np.linalg.norm(student_B)

cos_theta = dot_product / (magnitude_A * magnitude_B)
angle_radian = np.arccos(cos_theta)
angle_degree = np.degrees(angle_radian)

print("Angle in Radians:", angle_radian)
print("Angle in Degrees:", angle_degree)
```

### 5. Cross Product

```python
cross_product = np.cross(student_A, student_B)

print("Cross Product:", cross_product)
```

### 6. Vector Projection

```python
dot_product = np.dot(student_A, student_B)
magnitude_B = np.linalg.norm(student_B)

scalar_projection = dot_product / magnitude_B
vector_projection = (dot_product / (magnitude_B ** 2)) * student_B

print("Scalar Projection:", scalar_projection)
print("Vector Projection:", vector_projection)
```

### 7. Vector Visualization

```python
plt.quiver(0, 0, student_A[0], student_A[1], angles="xy", scale_units="xy", scale=1, label="Student A")
plt.quiver(0, 0, student_B[0], student_B[1], angles="xy", scale_units="xy", scale=1, label="Student B")
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.grid(True)
plt.title("Student Performance Vectors")
plt.xlabel("Subject 1")
plt.ylabel("Subject 2")
plt.legend()
plt.show()
```

```python
square_matrix = np.array([
    [1, 2],
    [3, 4]
])

determinant = np.linalg.det(square_matrix)
inverse = np.linalg.inv(square_matrix)

print("Determinant:", determinant)
print("Inverse:")
print(inverse)
```

### 14. Plane in 3D

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 20)
y = np.linspace(-5, 5, 20)

X, Y = np.meshgrid(x, y)
Z = 2 * X + 3 * Y + 1

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z)
ax.set_title("Plane in 3D")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```

## Part D: Eigenvalues & Decomposition

### 16. Eigenvalues and Eigenvectors

```python
A = np.array([
    [2, 1],
    [1, 2]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:")
print(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

### 17. Verify Av = λv

```python
v = eigenvectors[:, 0]
lambda_val = eigenvalues[0]

Av = A @ v
lambda_v = lambda_val * v

print("Av:", Av)
print("λv:", lambda_v)
```

### 18. Covariance Matrix

```python
data = np.array([
    [80, 75, 85],
    [65, 90, 78],
    [88, 82, 91],
    [70, 68, 75],
    [92, 88, 95]
])

covariance_matrix = np.cov(data, rowvar=False)

print("Covariance Matrix:")
print(covariance_matrix)
```

### 19. Eigenvalues and Eigenvectors of Covariance Matrix

```python
eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

### 20. Sorted Eigenvalues

```python
idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("Sorted Eigenvalues:")
print(eigenvalues)

print("\nSorted Eigenvectors:")
print(eigenvectors)
```

### 21. Explained Variance

```python
explained_variance = eigenvalues / np.sum(eigenvalues)

print("Explained Variance:")
print(explained_variance)

print("Explained Variance Percentage:")
print(explained_variance * 100)
```
